---
In this assignment, we compare two commonly used probability models: the Exponential and Poisson distributions. These models are fundamental in statistics for modeling real-world phenomena. The Exponential distribution is ideal for modeling the time between events in a continuous process, such as waiting times, while the Poisson distribution is used to model the number of events occurring in a fixed interval of time or space.

We selected these models because they are closely related—Poisson models the count of events, and Exponential models the time between those events—making them suitable for understanding patterns in event-based data. Comparing them allows us to analyze their fit to data, evaluate assumptions, and choose the most appropriate model for prediction or analysis.

##**1. THEORETICAL COMPARISON**




### **Write the probability density (or mass) function for each distribution. Identify and describe all parameters**

####**1. Poisson Distribution (discrete)**


**Probability mass function (PMF):**
$$
P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}, \quad k = 0,1,2,\ldots
$$
**Parameters:**

λ > 0:
- it's the rate (or mean) of events per fixed time interval.

It represents:
- Expected number of events in the interval
$$
E[X] = \lambda
$$

- Variance of the distribution
$$
\mathrm{Var}(X) = \lambda
$$

**Description:**
- it models the number of events occur in a fixed time interval and those events occur independently and at a constant average rate
- Discrete model


#### **2. Exponential Distribution (continuous)**
**Probability density function (PDF):**
$$
f(t) = \lambda e^{-\lambda t}, \quad t \ge 0
$$
**Parameters:**

λ > 0:
- Average number of events per unit time
- Inverse of the mean waiting time
$$
E[T] = \frac{1}{\lambda}
$$
$$
\mathrm{Var}(T) = \frac{1}{\lambda^2}
$$

**Description:**
- Models the waiting time until the next event
- Continuous-time model
- Has the memoryless property (*example:* If you already waited 10 minutes, the probability you must wait another 5 minutes is the same as the probability of waiting 5 minutes when you first arrived)

### **Report the expressions for the mean and variance. How do these quantities depend on the parameters ?**

#### **1. Poisson distribution:**
mean:
$$
E[X] = \lambda
$$
variance:
$$
\mathrm{Var}(X) = \lambda
$$

- Both mean and variance are equal to the same parameter λ
- So, increasing the parameter λ, increases linearly:
    1. the expected number of events
    2. the spread of the distribution

#### **2. Exponential distribution:**
mean:
$$
E[T] = \frac{1}{\lambda}
$$
variance:
$$
\mathrm{Var}(T) = \frac{1}{\lambda^2}
$$
- the mean is inversely proportional to λ
- the variance is inversely proportional to ${\lambda^2}$
- So, increasing the parameter λ:
    1. the expected waiting time gets shorter
    2. the variability in waiting times is less (waiting times are concetrated around the mean which means the distribution is less spread out, thus less variable)

### **Plot the distributions for different parameter values. How does changing each parameter affect the shape (e.g., symmetry, spread, skewness)?**

#### **Poisson distributions (different values of λ)**

```python
import numpy as np
import math
import matplotlib.pyplot as plt

# Values of lambda to compare
lambdas_poisson = [1, 4, 10]

# Range of k values
k = np.arange(0, 25)

for lam in lambdas_poisson:
    pmf = [(lam**i * math.exp(-lam)) / math.factorial(i) for i in k]

    plt.figure()
    plt.stem(k, pmf)
    plt.xlabel("k")
    plt.ylabel("P(X = k)")
    plt.title(f"Poisson Distribution (lambda = {lam})")
    plt.show()
```

λ=1 -> highly right skewed

λ=4 -> less skewed

λ=10 -> approximately symmetric

**Thus, we conclude that:**

in Poisson, increasing λ shifts the distribution right, increases spread, reduces skewness, and the distribution approaches symmetry.

#### **Exponential distributions (different values of λ)**

```python
# Values of lambda to compare
lambdas_exp = [0.5, 1, 3]

# Time axis
t = np.linspace(0, 10, 400)

for lam in lambdas_exp:
    pdf = lam * np.exp(-lam * t)

    plt.figure()
    plt.plot(t, pdf)
    plt.xlabel("t")
    plt.ylabel("f(t)")
    plt.title(f"Exponential Distribution (lambda = {lam})")
    plt.show()
```

small λ -> the distribution is highly spread, and the tail is long

large λ -> the peak gets sharper and gets close to 0, and the tail is shorter

And the distribution is **always** right skewed

**Thus, we conclude that:**

in Exponential, Increasing λ concentrates probability near zero, reduces spread, but skewness remains right-skewed because in exponential, the probability decreses as the time increases

### **Explain the practical meaning of the parameters in real-world contexts. When would one distribution be more suitable than the other?**

#### **1. Poisson distribution:**
λ: average number of events per fixed interval (time, area, length,...)

**Real-world examples:**
- Number of patients arriving at an emergency room per hour, suppose λ = 12, this means that: the average number of patients arriving to the emergency room per hour equals 12.
- Number of defects per meter of fabric
- Number of emails received per day

**Note:** λ here represents the mean and the variance,, and we use Poisson when we are counting events.

#### **2. Exponential distribution:**
λ: rate of events per unit time (expected number of events per unit time)

**Real-world examples:**
- Time until the next customer arrives
- Time until a machine fails
- Time until the next phone call

let's say we have λ = 2, this means:
- On average, 2 events per unit time
- mean waiting time = 1/2

**Note:** larger λ means that events happen more frequently, Thus, the waiting time becomes shorter.

**When to use each distribution ?**

* Use Poisson when:
  - we want to model the number of events
  - events occur independently and at a constant average rate.
  - the interval is fixed

* Use Exponential when:
  - we want to model time until an event
  - we care about waiting time
  - The process has no memory


### **Discuss the connection between your chosen distributions and the Gaussian probability model. Do they converge to the Normal distribution under specific circumstances? Do they relate to it through transformations or asymptotic behavior? Include formulas and visual examples where appropriate.**

The Poisson distribution converges to the Normal distribution as the rate parameter λ becomes large, with mean and variance both equal to
λ. The exponential distribution itself does not converge to a Normal distribution due to its skewness; however, the sum of independent exponential random variables converges to a Normal distribution as the number of terms increases, according to the **Central Limit Theorem** (the sum (or average) of many independent random variables becomes approximately Normal, regardless of their original distribution)

#### **In Poisson distribution:**

The mean = λ and the variance = λ, so as λ approaches ∞, the skewness decreases and the distribution becomes close to Normal.

**Visual example:**

suppose we have λ = 20, the Poisson PMF closely follows the bell-shaped Normal curve.



     
#### **In Exponential distribution:**

A single exponential random variable is always right-skewed and never converges to the normal distribution.
However, if we use the central limit theorem and sums many independent exponential variables, the result becomes approximately normal.

**Visual example:**

The histogram of the sum of 30 exponential waiting times is nearly symmetric and bell-shaped.

####**Poisson distribution visual example (large λ):**

```python
# Parameter
lam = 20

# Discrete values
k = np.arange(0, 40)

# Poisson PMF
poisson_pmf = [(lam**i * math.exp(-lam)) / math.factorial(i) for i in k]

# Normal approximation
normal_pdf = (1 / np.sqrt(2 * np.pi * lam)) * np.exp(-(k - lam)**2 / (2 * lam))

# Plot
plt.figure()
plt.stem(k, poisson_pmf)
plt.plot(k, normal_pdf)
plt.xlabel("k")
plt.ylabel("Probability / Density")
plt.title("Poisson Distribution (λ = 20) and Normal Approximation")
plt.show()
```

####**Exponential visual example (Sum of Exponential random variables):**

```python
# Parameters
lam = 1
n = 30          # number of exponentials to sum
N = 10000       # number of simulations

# Generate samples
samples = np.sum(np.random.exponential(1/lam, size=(N, n)), axis=1)

# Plot histogram
plt.figure()
plt.hist(samples, bins=50, density=True)
plt.xlabel("Sum of waiting times")
plt.ylabel("Density")
plt.title("Sum of Exponential Random Variables (Approx. Normal)")
plt.show()
```

### **Generate random samples (e.g., 1000 observations) from each of the two distributions using different parameter combinations, AND Plot histograms or/and density plots to visualize how the simulated data compare to the theoretical distributions.**

```python
# Fix seed for reproducibility
np.random.seed(42)

# Number of observations
n = 1000

# Poisson samples
poisson_samples_lambda_2 = np.random.poisson(lam=2, size=n)
poisson_samples_lambda_8 = np.random.poisson(lam=8, size=n)

# Exponential samples
exp_samples_lambda_05 = np.random.exponential(scale=1/0.5, size=n)
exp_samples_lambda_2 = np.random.exponential(scale=1/2, size=n)

# Combine into a DataFrame
samples = pd.DataFrame({
    "Poisson_lambda_2": poisson_samples_lambda_2,
    "Poisson_lambda_8": poisson_samples_lambda_8,
    "Exponential_lambda_0.5": exp_samples_lambda_05,
    "Exponential_lambda_2": exp_samples_lambda_2
})

# ------------------
# Poisson Histograms
# ------------------
poisson_lambdas = [2, 8]

for lam in poisson_lambdas:
    samples = np.random.poisson(lam=lam, size=n)

    plt.figure()
    plt.hist(samples, bins=range(min(samples), max(samples) + 2), density=True)
    plt.xlabel("k")
    plt.ylabel("Relative frequency")
    plt.title(f"Poisson Distribution Histogram (lambda = {lam})")
    plt.show()

# ----------------------
# Exponential Histograms
# ----------------------
exponential_lambdas = [0.5, 2]

for lam in exponential_lambdas:
    samples = np.random.exponential(scale=1/lam, size=n)

    plt.figure()
    plt.hist(samples, bins=30, density=True)
    plt.xlabel("t")
    plt.ylabel("Density")
    plt.title(f"Exponential Distribution Histogram (lambda = {lam})")
    plt.show()

# ------------------
# Compute sample mean and variance
# and compare with theoretical values
# ------------------

comparison = pd.DataFrame({
    "Distribution": [
        "Poisson (λ = 2)",
        "Poisson (λ = 8)",
        "Exponential (λ = 0.5)",
        "Exponential (λ = 2)"
    ],
    "Sample Mean": [
        poisson_samples_lambda_2.mean(),
        poisson_samples_lambda_8.mean(),
        exp_samples_lambda_05.mean(),
        exp_samples_lambda_2.mean()
    ],
    "Theoretical Mean": [
        2,
        8,
        1/0.5,
        1/2
    ],
    "Sample Variance": [
        poisson_samples_lambda_2.var(),
        poisson_samples_lambda_8.var(),
        exp_samples_lambda_05.var(),
        exp_samples_lambda_2.var()
    ],
    "Theoretical Variance": [
        2,
        8,
        1/(0.5**2),
        1/(2**2)
    ]
})

comparison
```

### **Comparison between the simulated data and the theoritical PDF/PMF:**

 A close comparison between the simulated data and the theoretical distributions shows strong agreement in both shape and summary statistics. This agreement improves with increasing sample size and is a direct consequence of the Law of Large Numbers (As the number of independent and identically distributed observations increases, the sample average converges to the theoretical (expected) value of the random variable)


### **Compute the sample mean and variance, and compare them with the theoretical values.**

As computed above in the code,, The sample mean and variance are very close to the theoretical mean and variance for both the Poisson and Exponential distributions. ### **Invent a context (realistic or hypothetical) in which the chosen distributions could model observable data.**
### **Clearly describe the nature of the random variable(s) involved, the assumption that justify the use of each distribution, the parameters of the model and how they could be interpreted.**

**Emergency room arrival**

**- Poisson Distribution (counts):**

let $X$ = number of patients arriving at the emergency room in 1 hour.

we can model this by a Poisson distribution because:
- arrivals occur independently
- the average arrival rate is approximately constant

Example:

let's say that we have arrival rate of 8 patients in 1 hour, this means that the parameter is $λ$ = 8:
$$
X \sim \text{Poisson}(\lambda = 8)
$$
So, we calculate the probability of exactly getting $k$ arrivals in 1 hour: $$P(X=k)$$

**- Exponential Distribution (waiting times):**
let $T$ = time in minutes between consecutive patient arrivals.

we can model this by an Exponential distribution because:
- waiting times are continuous
- the process has the memoryless property

Example:
as the previous example, we have ON average 8 events (arrivals) occur in 1 hour, So our parameter $λ$ = 8:
$$
T \sim \text{Exponential}(\lambda = 8 \text{ per hour})
$$
Thus, the **mean waiting time** will be:
$$
\mathbb{E}[T] = \frac{1}{\lambda} = \frac{1}{8}\ \text{hour} \approx 7.5\ \text{minutes}
$$




### **Simulate and analyze a dataset consistent with your scenario.**

```python
# -----------------------------
# Simulation setup
# -----------------------------
np.random.seed(42)

lambda_rate = 8   # average 8 patients per hour
hours = 24        # one full day

# -----------------------------
# Simulate the dataset
# -----------------------------

# Poisson: number of patients arriving per hour
arrivals_per_hour = np.random.poisson(lam=lambda_rate, size=hours)

# Exponential: time between patient arrivals (in hours)
interarrival_times = np.random.exponential(scale=1/lambda_rate, size=1000)

# Create datasets
poisson_data = pd.DataFrame({
    "Hour": range(1, hours + 1),
    "Arrivals_per_hour": arrivals_per_hour
})

exponential_data = pd.DataFrame({
    "Interarrival_time_hours": interarrival_times
})

# -----------------------------
# Analysis: mean and variance
# -----------------------------

analysis = pd.DataFrame({
    "Distribution": [
        "Poisson (λ = 8)",
        "Exponential (λ = 8)"
    ],
    "Sample Mean": [
        arrivals_per_hour.mean(),
        interarrival_times.mean()
    ],
    "Theoretical Mean": [
        lambda_rate,
        1 / lambda_rate
    ],
    "Sample Variance": [
        arrivals_per_hour.var(),
        interarrival_times.var()
    ],
    "Theoretical Variance": [
        lambda_rate,
        1 / (lambda_rate ** 2)
    ]
})

# Display results
print("Poisson dataset (arrivals per hour):")
print(poisson_data)

print("\nComparison of empirical and theoretical values:")
print(analysis)
```

This dataset was simulated to represent patient arrivals to a clinic assuming a constant arrival rate of $λ$=8 patients per hour. The number of arrivals per hour was modeled using a **Poisson distribution**, while the time between consecutive arrivals was modeled using an **exponential distribution**. The empirical mean and variance computed from the simulated data are close to the corresponding theoretical values predicted by the models.
